```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model = "gemini-3.6-flash", temperature=0.8)

response = llm.invoke("How many moons does Jupyter have?")
print(response.text)
```

```python
response = llm.invoke([
    ["system", "You are a helpful assistant that answers in short one line"],
    ["human", "How many moons does Jupiter have?"]
])

print(response.text)
```
